```python
from sysid.data_pipeline.data_interface import SysidDSInterface

ds = SysidDSInterface(dataset_name='dataset')
```

```python
import numpy as np
import matplotlib.pyplot as plt

rollout = ds.data['data'][10]
print(np.array(rollout['sensor_data']).shape)
print(np.array(rollout['actions']).shape)
print(np.array(rollout['velocities']).shape)
print(np.array(rollout['accelerations']).shape)
print(np.array(rollout['torque']).shape)

fig,axs = plt.subplots(4,1)
axs[0].plot(np.array(rollout['sensor_data']))
axs[1].plot(np.array(rollout['velocities']))
axs[2].plot(np.array(rollout['accelerations']))
axs[3].plot(np.array(rollout['torque']))

plt.show()
```
