```python
import json
import os

path = os.getcwd().split('/')[:-1]
filename = '/'.join(path) + '/dataset/actions.json'

with open(filename, 'r') as f:
    data = json.load(f)
```

```python
import matplotlib.pyplot as plt
import numpy as np

for i in range(10):
    actions = np.array(data[i]['actions'])
    plt.plot(actions)
    plt.show()
```
